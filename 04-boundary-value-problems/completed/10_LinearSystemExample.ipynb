## Linear system solution

$$ 1a+1b = 35$$
$$2a+4b=94$$

$$A=\begin{bmatrix} 
      1 & 1 \\
      2 & 4 \\
\end{bmatrix}$$

$$X=\begin{bmatrix} 
      a \\
      b \\
\end{bmatrix}$$

$$B=\begin{bmatrix} 
      35 \\
      94 \\
\end{bmatrix}$$

$$AX=B$$

or

$$ \begin{bmatrix} 
      1 & 1 \\
      2 & 4 \\
\end{bmatrix}\begin{bmatrix} 
      a \\
      b \\
\end{bmatrix} =\begin{bmatrix} 
      35 \\
      94 \\
\end{bmatrix}$$

So, we need to invert $A$ to solve for $X$

$$ A^{-1} AX = A^{-1} B \\ X = A^{-1} B $$


```python
# Solving following system of linear equation
# 1a + 1b = 35
# 2a + 4b = 94

a = np.array([[1, 1],[2,4]])
b = np.array([35, 94])
x = np.linalg.solve(a,b)

print(x)
```

```python
# the inverse of a, then X = A^-1 B using @ (matrix multiplication)
a_inverse = np.linalg.inv(a)
xx = a_inverse @ b
print(a_inverse)
print(xx)
```

```python
# careful: * on NumPy arrays is element-wise multiplication, NOT matrix multiplication
# this does not give the solution:
a_inverse * b
```

```python
# Defining the matrices as 2D arrays (np.matrix is no longer recommended)
A = np.array([[1, 1], [2, 4]])
B = np.array([[35], [94]])

# Taking inverse of A
A_inverse = np.linalg.inv(A)

# Solving for X
X = A_inverse @ B
print(A_inverse)
print(X)

# check: A @ X should reproduce B
print(A @ X)
# http://docs.scipy.org/doc/numpy/reference/routines.linalg.html
```
