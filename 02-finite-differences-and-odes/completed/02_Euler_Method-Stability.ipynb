# Euler Method - Stability

Solve the following equation:

\begin{equation}
\frac{d y}{dx}=-100y
\end{equation}

using and explict Euler method. Solve it from $x=0$ to $x=0.3$. Take the initial condition as being $y(x=0)=1$.

## Global input

```python
# input the needed conditions
start = 0
end = 0.3
y0 = 1
```

## The analytic solution

```python
# define the array
x_a = np.linspace(start,end,100)
y_a = y0*np.exp(-100*x_a)

# plot the analytic solution
fig, ax = plt.subplots()
ax.plot(x_a, y_a) 
ax.set_xlabel('$x$')
ax.set_ylabel('$y$');
```

## Explicit numeric solution

define the function $f(x,y)$

```python
def f(y):
    return -100*y
```

The new input needed is the value of $\Delta x$

```python
# input
dx = 0.012

# setup the array... use arange and dx for x and zeros for y
x = np.arange(start,end,dx)
N = len(x) # length of the t array

y = np.zeros(N)
y[0]= y0 # the initial condition

# calculate y numerically
for i in range(0,N-1):
    y[i+1]=y[i]+dx*f(y[i])
```

```python
# plot the analytic and numeric solution
fig, ax = plt.subplots()
ax.plot(x, y, '-o', label='Euler') 
ax.plot(x_a, y_a, label='Analytic') 

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
plt.legend();
```

## Implicit numeric solution

```python
# input
dx = 0.1
l = 100

# setup the array... use arange and dx for x and zeros for y
x = np.arange(start,end,dx)
N = len(x) # length of the t array

y = np.zeros(N)
y[0]= y0 # the initial condition

# calculate y numerically
for i in range(0,N-1):
    y[i+1]=(1/(1+l*dx))*y[i]

# plot the analytic and numeric solution
fig, ax = plt.subplots()
ax.plot(x, y, '-o', label='Euler') 
ax.plot(x_a, y_a, label='Analytic') 

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
plt.legend();
```
